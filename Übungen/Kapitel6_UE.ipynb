Wo ist das Problem bei dieser Function, die eine csv Datei einlesen und als Liste von Listen zurückgeben soll?

```python
def read_labor_market_file(filename):
    all_incomes = []

    with open(filename, mode="r") as file:
        reader = csv.reader(file)
        for row in reader:  
            year_incomes = []
            for entry in row:
                income = float(row)
                year_incomes.append(income)
            all_incomes.append(year_incomes)
    return all_incomes
```

- Es sollte `year_incomes` returned werden
