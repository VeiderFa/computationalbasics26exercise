**Aufgabe 3**  
Wir schreiben ein Modell, das eine Hasenpopulation darstellen soll. Diese wächst immer schneller: Im ersten Zeitschritt kommen 2 Hasen dazu, dann 4, dann 8, dann 16 und so weiter. Simuliere 10 Zeitschritte und printe die resultierende Liste. Hier der Anfang des Programms:



```python
hasen = 20
hasen_liste = [20]
hasen_dazu = 2

for it in range(10): # it geht von 0 bis 9, also 10 Zeitschritte werden simuliert
    hasen = hasen + hasen_dazu   
    # hasen += hasen_dazu # Alternative Zeile, weil x = x + a ist gleich wie x += a
    hasen_liste.append(hasen) 
    hasen_dazu = hasen_dazu * 2 
    # hasen_dazu *= 2 # Alternative Zeile, weil x = x * a ist gleich wie x *= a
print(hasen_liste)
# Es gibt auch -=, /= für Subtraktion und Division
```

**Aufgabe 5**  
Erstelle eine Simulation, die eine Population von Lemmingen simuliert. Sie startet bei 1000. In jedem Zeitschritt wird die Zahl der Lemminge zuerst halbiert und dann kommen 50 Lemminge dazu. Simuliere 15 Zeitschritte und erstelle einen Plot. Was ändert sich, wenn wir stattdessen mit 10 Lemmingen (statt mit 1000) starten würden?

```python
import matplotlib.pyplot as plt

# Bei 10 Lemmingen steigt die Anzahl bis 100, bei 1000 fällt sie auf 100
# Die Anzahl pendelt sich genau bei 100 ein, es handelt sich um einen "Stationären Zustand" ("steady state")
# Mehr dazu in anderen Kursen

lemminge = 10 
lemming_liste = [lemminge]

for it in range(15):
    lemminge = lemminge / 2 + 50
    lemming_liste.append(lemminge)

print(lemming_liste)

plt.plot(lemming_liste)
plt.xlabel("Zeit") # Optionale Beschreibung der x-Achse, mehr davon in folgenden Einheiten
plt.ylabel("Anzahl Lemminge") # Optionale Beschreibung der y-Achse
plt.show()
```

Füge eine Zeile zum unten stehenden Programm hinzu. Am Schluss soll folgende Liste geprintet werden: `[10, 30, 90, 270, 810]`

```python
# Der Trick ist zu sehen dass sich die Zahlen immer verdreifachen
# Wenn nicht gefordert wird dass man eine Schleife verwendet könnte man auch einfach
# liste = [10, 30, 90, 270, 810] händisch tippen. 
# Wenn aber eine Liste generiert werden soll mit z.B. 100+ Einträge 
# dann ist händisches Eintragen schon aufwendig. 

zahl = 10 
liste = [zahl]

for it in range(4): 
    zahl = zahl * 3 
    liste.append(zahl)

print(liste)
```
