## Abschluss-Frage
Welcher dieser Funktionsaufrufe wird eine Fehlermeldung verursachen?

```python
def rechnungteilen(betrag,personen,trinkgeld = 1.1):
    return betrag * trinkgeld / personen
```

- `rechnungteilen(100,10)`
- `rechnungteilen(10,100)`
- `rechnungteilen(150,12,trinkgeld = 0.9)`
- `rechnungteilen(100,trinkgeld = 1.1)`
