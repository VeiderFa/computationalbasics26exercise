## Aufgabe 5
Verbessere den scatter plot: Die Punkte, die die häufigste Farbe haben, sollen größer gezeichnet werden (`s=100`). Wenn man das ganz sauber programmieren will, sollte man zuerst die Liste mit Punkten erstellen, dann einmal die häufigste Farbe ausrechnen und diese Farbe dann als Argument für die zeichen-Methode mitgeben. Tipp für alle, die mit einem Dictonary arbeiten: `max(dict)` liefert den alphabetisch größten Key. Wenn man den Key haben will, der den größten Wert hat, kann man `max(dict,key=dict.get)` verwenden.

```python
#-------------
#Musterlösung
#-------------

import random
import matplotlib.pyplot as plt

class Punkt():
    def __init__(self):
        self.x=random.uniform(0,100)
        self.y=random.uniform(0,100)
        r = ((self.x-50)**2 + (self.y-50)**2)**0.5 
        if r < 10:
            c=0
        elif r < 20:
            c=1
        elif r < 40:
            c=2
        else:
            c=3
        self.color=["crimson","forestgreen","skyblue","darkorchid"][c]
    def zeichnen(self,maxi):
        size = 50
        if self.color==maxi:
            size=100
        plt.scatter(self.x,self.y,color=self.color,s=size)


def count(liste):
    colors={}
    for p in liste:
        c=p.color
        if c in colors:
            colors[c]+=1
        else:
            colors[c]=1
    return colors


    
punkte=[]
for it in range(500):
    punkte.append(Punkt())

di = count(punkte)

maxi = max(di,key=di.get)

for p in punkte:
    p.zeichnen(maxi)
plt.show()

```
